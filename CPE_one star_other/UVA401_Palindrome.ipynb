# Palindrome

迴文字串是一串數字或字母，從左到右讀取和從右到左讀取相同。 例如，”ABCDEDCBA”是迴文字串。 鏡像字串是一種字串，當該字串的每個元素更改為鏡像(如果它具有鏡像)並且從右到左讀取該字串時，其結果與從左到右讀取原始字串相同。 例如，”3AIAE”是鏡像字符串，因為”A”的鏡像和”I”的鏡像是他們自己，而”3”和”E”為彼此的鏡像。 鏡像迴文是指符合迴文字串標準和鏡像字串標準的字串。 例如，”ATOYOTA”是一個鏡像迴文，”A”、”T”、”O”、”Y”為彼此的鏡像。 該字串從左到右讀取和從右到左讀取相同。 並且每個字元都鏡像替換後從右到左讀取結果，與從左到右讀取原始字串相同。





```python
import sys

mirror = {'A':'A', 'E':'3', 'H':'H', 'I':'I', 'J':'L', 'L':'J', 'M':'M', 'O':'O', 'S':'2',
		  'T':'T', 'U':'U', 'V':'V', 'W':'W', 'X':'X', 'Y':'Y', 'Z':'5', '1':'1', '2':'S',
		  '3':'E', '5':'Z', '8':'8'}

		 
def is_palindrome(s, s_half_length): #s will be the string
	for i in range(0, s_half_length):
		if s[i] != s[-1-i]:
			return False
	return True

	
def is_mirrored(s, s_half_length):
	for i in range(0, s_half_length):
		if s[i] not in mirror:
			return False
		elif s[-1-i] != mirror[s[i]]:
			return False
	if len(s) % 2 == 1: #discuss the middle string
		if s[s_half_length] not in mirror:
			return False
		elif s[s_half_length] != mirror[s[s_half_length]]:
			return False
	return True


for line in sys.stdin: #line is string
	line = line.strip() #remove the \n \r which regarded as string
	L =  len(line) #lenth of the string
	L_half = L // 2 #half of the length
	a, b = is_palindrome(line, L_half), is_mirrored(line, L_half)
	
	if a and b:
		print(line,"-- is a mirrored palindrome.\n")
	elif a:
		print(line,"-- is a regular palindrome.\n")
	elif b:
		print(line,"-- is a mirrored string.\n")
	else:
		print(line,"-- is not a palindrome.\n")
```

### Thinkthrough

大架構來說，分兩塊來判斷：

1. 是否為 palindrome
2. 是否為 mirrored

兩種情形恰好可以用布林值來呈現四種情形。

注意細節：
    
1. 建立字典來對應鏡像關係，字典中的對應皆是單方向，比如 'E':'3' 便是由 E 至 3。
2. 字典取值要用中括號，小括號電腦會認為要取函數而非取值。
3. def的參數不限型態，只要對應好即可，而return的部分，但凡迴圈內return成功，便會離開def。
4. for line in sys.stdin去讀取標準輸入，要注意換行符號全部也會被當作字串，可用strip處理。
5. 判斷非鏡像，要先考慮字元有沒有在字典裡，有的話才去判斷是否具有鏡像關係，否則迴圈會卡住。
6. a, b是布林值，可以直接從他們組合思考，用if else去做排除篩選。
7. not in mirror注意不要加is，英文文法不適用。
