### Divide, But Not Quite Conquer!


```python
### 自寫 ###
import sys

for line in sys.stdin:
	n, m = map(int, line.split())
	if m <= 1 or n <= 1:
		print("Boring!")
		continue
	
	seq = [n]
	while n > 0:
		q, r = divmod(n, m)
		if r == 0:
			if q == 1:
				seq.append(q)
				print(*seq)
				break
			else:
				seq.append(q)
				n = q
		else:
			print("Boring!")
			break
```

Note:
1. print(*seq)用法下方會特別說明。此題其他寫法有:
    - print(" ".join(map(str, seq)))
    - print(" ".join(str(i) for i in seq))
    - for迴圈見下方範例。

```python
### print(*seq)此題的其他寫法 ###
for i in range(len(seq)):
    if i > 0:
        print(" ", end="")
    print(seq[i], end="")

print()


### 關於print(*seq) ###
# 不限定 seq 必須是 list，也不要求裡面的元素一定要是數字或字串。
# *seq 的作用是將可迭代物件（iterable）拆開，作為多個引數傳入 print()。

## 範例 ## 
seq = [8, 4, 2, 1]
print(*seq)
# 相當於: 
print(8, 4, 2, 1)

## 1. seq 可以是其他可迭代型態 ##
print(*(8, 4, 2, 1))     # tuple：8 4 2 1
print(*range(4))          # range：0 1 2 3
print(*"Hello")           # 字串：H e l l o

# 字典也可以，但預設拆出的是 key：
seq = {"a": 10, "b": 20}
print(*seq)               # a b

# 單一整數不能直接拆開，因為它不是可迭代物件：
seq = 123
print(*seq)              # TypeError

## 2. 元素可以混合不同型態 ##
seq = [10, "apple", 3.5, True, None]
print(*seq)
# 輸出: 10 apple 3.5 True None

# 因為 print() 會自動將各個引數轉成文字，所以不需要像 join() 一樣，先將所有元素轉成字串。
# 也能用 sep 指定分隔方式:
print(*seq, sep=", ")
# 輸出: 10, apple, 3.5, True, None
```
