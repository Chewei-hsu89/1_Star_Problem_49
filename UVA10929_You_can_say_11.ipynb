# You can say 11

Your job is, given a positive number N, determine if it is a multiple of eleven. 

Input 

The input is a file such that each line contains a positive number. A line containing the number ‘0’ is the end of the input. The given numbers can contain up to 1000 digits. 


Output 

The output of the program shall indicate, for each input number, if it is a multiple of eleven or not. 


Sample Input 

112233

30800

2937

323455693

5038297

112234

0



Sample Output 

112233 is a multiple of 11.

30800 is a multiple of 11.

2937 is a multiple of 11.

323455693 is a multiple of 11.

5038297 is a multiple of 11.

112234 is not a multiple of 11.

```python
while True:
	N = int(input())
	M = N
	if N == 0:
		break
	
	digit = []
	while N > 0:
		# 商數餘數可以用 N, R = divmod(N, 10)直接解決
		R = N % 10
		N = N // 10 
		digit.append(R)
	
	sum_even, sum_odd = 0, 0	
	for i in range(len(digit)):
		if i % 2 == 0:
			sum_even += digit[i]
		else:
			sum_odd += digit[i]
			
	if abs(sum_even - sum_odd) % 11 == 0:
		print(M, "is a multiple of 11.")
	else:
		print(M, "is not a multiple of 11.")
```

```python
while True:
    number = input().strip() #字串型態

    if number == "0":
        break

    sum_even, sum_odd = 0, 0

    for i, digit in enumerate(number):
        if i % 2 == 0:
            sum_even += int(digit)
        else:
            sum_odd += int(digit)

    if abs(sum_even - sum_odd) % 11 == 0:
        print(number, "is a multiple of 11.")
    else:
        print(number, "is not a multiple of 11.")
```
