# The 3n+1 Problem

Algorithm:
1. input n
2. print n
3. if n = 1 then STOP
4. if n is odd then n ← 3n + 1
5. else n ← n/2
6. GOTO 2

ex. 22 11 34 17 52 26 13 40 20 10 5 16 8 4 2 1 
據推測此演算法對任何整數而言會終止 (當列印出 1 的時候)。雖然此演算法很簡單，但以上的推測是否真實卻無法知道。然而對所有的n ( 0 < n < 1,000,000 )來說，以上的推測已經被驗證是正確的。 

給一個輸入 n ,透過以上的演算法我們可以得到一個數列（1作為結尾）。此數列的長度稱為n的cycle-length。上面提到的例子, 22 的 cycle length為 16. 

問題來了：對任2個整數i，j我們想要知道介於i，j（包含i，j）之間的數所產生的數列中最大的 cycle length 是多少。

|輸入說明 |輸入可能包含了好幾列測試資料，每一列有一對整數資料 i，j 。 0< i，j < 1,000,000|
|---|---|
|輸出說明|對每一對輸入 i , j 你應該要輸出 i, j 和介於 i, j 之間的數所產生的數列中最大的 cycle length。|

-----------

|Sample Input | Sample Output|
|---|---|
|1 10<br>10 1<br>100 200<br>201 210<br>900 1000 |1 10 20<br>10 1 20<br>100 200 125<br>201 210 89<br>900 1000 174|

Thinkthrough:

先用單獨寫出演算法函式，其輸出為週期長度。然後讀取兩個數之間，算出他們的值並比較出最大值，注意兩個數的大小順序，可能會影響for迴圈的運作。

```python
import sys

def cycle_algorithm(n):
	cyc_length = 1
	
	while n != 1:
		if n % 2 == 1:
			n = 3*n + 1
			cyc_length += 1
		else:
			n = n/2 # 非整除法，會有浮點數
			cyc_length += 1
			
	return cyc_length

	
for line in sys.stdin:
	a, b = map(int, line.split())
	if a > b:
		c, d = b, a
	else:
		c, d = a, b	
	
	max_cyc_length = cycle_algorithm(c)
	
	for i in range(c, d+1):
		if cycle_algorithm(i) > max_cyc_length:
			max_cyc_length = cycle_algorithm(i)
			
	print(a, b, max_cyc_length)
```

```python
#精修版
import sys


def cycle_algorithm(n):
    cycle_length = 1

    while n != 1:
        if n % 2 == 1:
            n = 3 * n + 1
        else:
            n = n // 2 #整除法

        cycle_length += 1

    return cycle_length


for line in sys.stdin:
    a, b = map(int, line.split())

    c, d = min(a, b), max(a, b)

    max_cycle_length = 0

    for i in range(c, d + 1):
        max_cycle_length = max(max_cycle_length, cycle_algorithm(i))

    print(a, b, max_cycle_length)
```
