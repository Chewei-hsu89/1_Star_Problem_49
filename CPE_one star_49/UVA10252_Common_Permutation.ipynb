# Common Permutation

Time Limit: 4 sec

Description

Given two strings of lowercase letters, a and b, print the longest string x of lowercase letters such that there is a permutation of x that is a subsequence of a and there is a permutation of x that is a subsequence of b.

Input

Input file contains several lines of input. Consecutive two lines make a set of input. That means in the input file line 1 and 2 is a set of input, line 3 and 4 is a set of input and so on. The first line of a pair contains a and the second contains b. Each string is on a separate line and consists of at most 1000 lowercase letters.

Output

For each set of input, output a line containing x. If several x satisfy the criteria above, choose the first one in alphabetical order.

--------------------------------------------------------------------------------



```python
import sys

while True:
	try:
		a = input()
		b = input()

		x_letter = set(a).intersection(set(b))
		x_letter.discard(" ")
		x_letter = list(x_letter)
		x_letter.sort()
		x_all = []
		for i in x_letter:
			for j in range(min(a.count(i), b.count(i))):
				x_all.append(i)
		print("".join(x_all))
		
	except EOFError:
		break
```

```python
# 修正精進版
while True:
    try:
        a = input()
        b = input()

        common_letters = sorted(set(a).intersection(b))
        common_letters.discard(" ")
        result = []

        for letter in common_letters:
            amount = min(a.count(letter), b.count(letter))
            result.append(letter * amount)

        print("".join(result))

    except EOFError:
        break
```

注意：
1. 若用input()，a, b是字串型態，後續會一個一個letter去做處理，若為了在此處處理" "而加入split()，會變list型態，會被視為一個單字存在list裡面，而失去更簡便的逐一處理模式。
2. set(a).intersection(b)這邊b可先轉set或不轉也行，intersection的動作會自動轉。
3. 用sort會依Unicode由小到大排列(ASCII順序也包含在Unicode內)，而.sort()不會回傳值，因此要獨立寫一行x_letter.sort()，若用sorted(x)則會回傳值，要在用一個變數來儲存。
4. .count()可以用來計數，會回傳值。

```python
# 用sys版本來讀取標準輸入
import sys

lines = sys.stdin.read().splitlines()

for i in range(0, len(lines), 2):
    a = lines[i]
    b = lines[i + 1]

    common_letters = set(a).intersection(b)
    common_letters.discard(" ")
    common_letters = sorted(common_letters)

    result = []

    for letter in common_letters:
        amount = min(a.count(letter), b.count(letter))
        result.append(letter * amount)

    print("".join(result))
```

補充:
1. splitlines()一樣是回傳list，但是依換行符號做切割。
2. read()是一次讀所有的內容直到EOF，回傳型態為str。
3. 因為做完splitlines()回傳list，所以要另外用for迴圈將每組資料提出。
