# Primary Arithmetic

Time Limit: 3 sec

Description

Children are taught to add multi-digit numbers from right-to-left one digit at a time. Many find the "carry" operation - in which a 1 is carried from one digit position to be added to the next - to be a significant challenge. Your job is to count the number of carry operations for each of a set of addition problems so that educators may assess their difficulty.

Input

Each line of input contains two unsigned integers less than 10 digits. The last line of input contains 0 0.

Output

For each line of input except the last you should compute and print the number of carry operations that would result from adding the two numbers, in the format shown below.

--------------------------------------------------------------------------------

|Sample Input | Sample Output|
|---|---|
|123 456<br>555 555<br>123 594<br>0 0|No carry operation.<br>3 carry operations.<br>1 carry operation.<br><br>|
 

 


### Thinkthrough:

題目求總進位次數，可以從每個位數的進位情況處理，先一路從個位數、十位數...往下推演，每次推演可以除以10，由餘數與前一位數進位的數字加總判斷。

```python
import sys

from io import StringIO
raw_input = """123 456
555 555
123 594
0 0
"""
sys.stdin = StringIO(raw_input)

for line in sys.stdin:
	if line.strip() == "0 0":
		break
		
	a, b = map(int, line.split())
	
	carry = 0 # store total carry time to count
	c = 0 # store carry for next digit
	while a > 0 or b > 0:
		ra = a % 10
		rb = b % 10
		
		if ra + rb + c > 9:
			c = 1
			carry += 1
		else:
			c = 0
		
		a //= 10
		b //= 10
			
	if carry == 0:
		print("No carry operation.")
	elif carry == 1: # 注意python沒有else if的寫法，只有elif
		print(carry, "carry operation.")
	else:
		print(carry, "carry operations.")
```
