```python
### 自寫 ###
import sys

def div3(x):
	num = 0
	while x > 0:
		q, r = divmod(x, 3)
		num += 3*q
		x = q + r
		if x == 2:
			num += 3
			return num
		if x == 1:
			num += 1
			return num

for line in sys.stdin:
	bottle = int(line)
	print(div3(bottle))
```

```python
### 速解: 3/2*n (直接背)###
import sys

for line in sys.stdin:
	bottle = int(line)
	print(int(3/2*bottle))
```

Note:
1. 我的演算法邏輯: 三個一組喝掉，喝掉便計數，換來的瓶子與剩下無法成三的瓶子加總，繼續判斷能否成三個一組，最後全部無法再成組，再處理剩餘瓶子。
2. return之後就不必加break，因為return完會直接跳脫整個函式。
