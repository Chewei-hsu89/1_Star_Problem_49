# Parity


```python
while True:
	I = int(input())
	if I == 0:
		break
	parity = 0
	bit = []
	while I >= 1:
		B = I % 2
		bit.append(B)
		if B == 1:
			parity += 1
		
		I = I // 2
	bit.reverse() #將list中的排列反轉
	print(f"The parity of {''.join(map(str, bit))} is {parity} (mod 2).") #利用空字串去將字串做join
```

注意：
1. input()讀出來的數字為字串，但已經將尾巴的迴車都去除了。
2. I, = map(int, input())的寫法也可以，要記得逗號，若單純只有I，會將I判斷為map物件，而沒有將映射的值取出。
3. .append()、.reverse()皆是獨立一行的動作，不用再拿變數去存，會出錯。
4. list要轉成數字表達，建議轉成字串透過join連接，當然其他方法也可以，比如用for迴圈去串接數字或字串，或用解包*。
5. 不想用reverse也可以用bit.insert(0, B)
6. ''.join(reversed(bit))則可以少一行bit.reverse()。

```python
#精修版
while True:
    I = int(input())

    if I == 0:
        break

    parity = 0
    bit = []

    while I > 0:
        B = I % 2
        parity += B
        bit.append(B)
        I //= 2

    bit.reverse()

    print(f"The parity of {''.join(map(str, bit))} is {parity} (mod 2).")
```

補充:
| 方法          | 修改原物件    | 有回傳值   | 建議                 |
| ----------- | -------- | ------ | ------------------ |
| `append()`  | ✔        | `None` | 獨立一行               |
| `reverse()` | ✔        | `None` | 獨立一行               |
| `sort()`    | ✔        | `None` | 獨立一行               |
| `join()`    | ✘        | 新字串    | 可直接放在 `print()`    |
| `strip()`   | ✘（字串不可變） | 新字串    | 要接回變數              |
| `replace()` | ✘        | 新字串    | 要接回變數              |
| `split()`   | ✘        | 新 list | 可直接接 `map()` 或指定變數 |
