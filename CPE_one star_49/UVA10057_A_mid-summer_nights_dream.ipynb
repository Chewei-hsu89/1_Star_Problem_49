For each set of input there will be one line of output. That line will contain the minimum possible value for A. Next it will contain how many numbers are there in the input that satisfy the property of A (The summation of absolute deviation from A is minimum). And finally you have to print how many possible different integer values are there for A (these values need not be present in the input). These numbers will be separated by single space.

意即:
- 第一個數字: 使得該式為最小的A值，選最小的A。
- 第二個數字: input中使該式為最小的A，計算其個數，其中可能會有重複數值。
- 第三個數字: 滿足該式的A值區間所有個數。

```python
### 自寫 ###

import math

while True:
	try:
		T = int(input())
		nums = []
		for _ in range(T):
			nums.append(int(input()))
		nums.sort()
		
		first_num = nums[math.ceil(T/2)-1]
		
		if T % 2 == 0:
			if first_num == nums[(T//2)]:
				second_num = nums.count(first_num)
				third_num = 1
			else:
				second_num = 0
				second_num = nums.count(first_num) + nums.count(nums[(T//2)])
				 
				third_num = 0
				for i in range(first_num, nums[(T//2)]+1):
					third_num += 1
				
		else:
			second_num = nums.count(first_num)
			third_num = 1
			
		print(f"{first_num} {second_num} {third_num}")
	
	except EOFError:
		break
```

```python
### 精修版 ###

while True:
    try:
        n = int(input())

        nums = []

        for _ in range(n):
            nums.append(int(input()))

        nums.sort()

        ## 直接用lower、upper median來處理更簡潔
        lower_median = nums[(n - 1) // 2]
        upper_median = nums[n // 2]

        if lower_median == upper_median:
            data_count = nums.count(lower_median)
        else:
            data_count = nums.count(lower_median) + nums.count(upper_median)

        answer_count = upper_median - lower_median + 1

        print(lower_median, data_count, answer_count)

    except EOFError:
        break
```

```python
while True:
    try:
        n = int(input())
        nums = [int(input()) for _ in range(n)]
        nums.sort()

        lower_median = nums[(n - 1) // 2]
        upper_median = nums[n // 2]

        data_count = sum(lower_median <= number <= upper_median for number in nums) #用list comprehension更簡潔

        answer_count = upper_median - lower_median + 1

        print(lower_median, data_count, answer_count)

    except EOFError:
        break
```
