```python
import math

T = int(input())

for i in range(T):
	lines = input().split()
	lines = list(map(int, lines))
	num = lines[0] #quantity of relation
	del lines[0] #移除第一項
	lines.sort()
	home = lines[math.ceil(num/2)-1] #值為中間偏左
	
	sum = 0 #名稱盡量不要用sum，會導致內建函數sum被覆蓋
	for j in lines:
		sum += abs(j-home)
	print(sum)
```

```python
T = int(input())

for _ in range(T):
    data = list(map(int, input().split())) #合併自寫的兩行

    relative_count = data[0]
    addresses = data[1:] #利用slice將第一個index內容切除，data可保留原始樣貌
    addresses.sort()

    home = addresses[relative_count // 2] #值會是中間偏左或偏右，不一定是最小的，但對距離計算沒影響

    total_distance = 0

    for address in addresses:
        total_distance += abs(address - home)

    print(total_distance)
```

### Note
1. list內部型態由字串轉數字:
    - list(map(int, lines))
    - numbers = [int(number) for number in numbers]
2. 移除的方法:(下方有範例測試)
    - 若知元素值: .remove(元素值)，該元素會移除，但不會回傳值。另外，若元素值有重複，則會先刪除第一個。
    - 若知索引: .pop(索引值)，該索引對應的元素會被移除，並回傳被移除的元素值。
    - 若知所引但不要回傳值: del numbers[索引]
    - 移除所有指定值: list comprhension 如 numbers = [number for number in numbers if number != 20]。
    - 清空整個list: .clear()
3. sort筆記:
    - numbers.sort(): 直接修改list的順序。
    - result = sorted(numbers): sorted會回傳修改後的list樣態，但numbers本身沒有改變。

```python
## 移除之範例測試 ##
numbers = [10, 20, 30, 40, 50]

print(numbers.pop(1))
print(numbers)

numbers.remove(10)
print(numbers)

del numbers[1]
print(numbers)

numbers.clear()
print(numbers)
```
