# Decode the Mad Man


```python
r = {'e':'q', 'd':'a', 'c':'z', 'r':'w', 'f':'s', 'v':'x', 't':'e', 'g':'d', 'b':'c', 'y':'r',
	 'h':'f', 'n':'v', 'u':'t', 'j':'g', 'm':'b', 'i':'y', 'k':'h', ',':'n', 'o':'u', 'l':'j',
	 '.':'m', 'p':'i', ';':'k', '/':',', '[':'o', "'":'l', ']':'p', '2':'`', '3':'1', '4':'2',
	 '5':'3', '6':'4', '7':'5', '8':'6', '9':'7', '0':'8', '-':'9', '=':'0', '\\':'[', ' ':' '}

while True:
	try:
		line = input()
		
		letter = []
		for i in line:
			letter.append(r[i])
		print("".join(letter))
		
	except EOFError:
		break
```

注意:
1. dictionary中，\為特殊字元，要寫成\\
2. '外側要用"包起來，不能單寫為'''
3. while迴圈遇到EOF要寫try、except
4. for i in line:可以直接將字串一個一個讀取

```python
#改良版，r的對應這邊暫時省略
import sys

for line in sys.stdin:
    line = line.rstrip("\n") #只移除右側的換行字元
    print("".join(r.get(i, i) for i in line))
```
