### All You Need Is Love



【題意精要】
判斷是否能夠找到一個合法的L，使得S1和S2都可以用Love做成。亦即
- S1 是 L 的倍數。
- S2 是 L 的倍數。
- L 是 S1 和 S2 的公因數。

```python
import math

T = int(input())

for i in range(T):
	S1 = int(input(), 2)
	S2 = int(input(), 2)
	
	L = math.gcd(S1, S2)
	
	if L != 1:
		print(f"Pair #{i+1}: All you need is love!")
	else:
		print(f"Pair #{i+1}: Love is not all you need!")
```

Note:
- math.gcd()為最大公因數
- math.lcm()為最小公倍數
- int(input(), 2)為二進位轉十進位寫法，特別注意input()必須是符合二進位的「字串」，而非數字型態。
