10424:Satellites

Time Limit: 3 sec

Description

The radius of earth is 6440 Kilometer. There are many Satellites and Asteroids moving around the earth. If two Satellites create an angle with the center of earth, can you find out thedistance between them? By distance we mean both the arc and chord distances. Both satellites are on the same orbit. (However, please consider that they are revolving on a circular path rather than an elliptical path.)



Input

The input file will contain one or more test cases.

Each test case consists of one line containing two-integer $s$ and $a$ and a string "min" or "deg". Here $s$ is the distance of the satellite from the surface of the earth and $a$ is the angle that the satellites make with the center of earth. It may be in minutes ( 嚙踝蕭 ) or in degrees ( 0 ). Remember that the same line will never contain minute and degree at a time.

Output

For each test case, print one line containing the required distances i.e. both arc distance and chord distance respectively between two satellites in Kilometer. The distance will be a floating-point value with six digits after decimal point.


--------------------------------------------------------------------------------

|Sample Input|Sample Output|
|------------|-------------|  
|500 30 deg<br> 700 60 min<br> 200 45 deg|3633.775503 3592.408346<br> 124.616509 124.614927<br> 5215.043805 5082.035982|



知識背景補充:
1. - arc distance: 弧長
    - chord distance: 弦長(兩個衛星之間的距離)
2. - Degrees (°), minutes (‘), seconds (“)
    - 1° = 60′
    - 1′ = 60″
3. 弦長 chord = 2* r * cos((90 – a / 2) / 180 * PI) = 2*(r + s)* sin(a/2)

```python
import sys
import math

for line in sys.stdin:
	s, a, m = line.split() # str type 沒必要再弄 map(str, ...)
	s = float(s)
	a = float(a)* math.pi / 180 #turn into radian
	if m == "min":
		a = a / 60
	
	arc_d = (6440 + s) * a
	chord_d = 2*(6440 + s)* math.sin(a/2)
	print(f"{arc_d:.6f} {chord_d:.6f}") # 注意小數點位數寫法
```

```python
import sys
import math

EARTH_RADIUS = 6440

for line in sys.stdin:
    altitude, angle, unit = line.split()

    altitude = float(altitude)
    angle = float(angle)

    if unit == "min":
        angle /= 60

    angle %= 360
    angle = min(angle, 360 - angle) # 考量到大角度問題(超過180度)，取小者，但測資沒有，不影響。

    radius = EARTH_RADIUS + altitude
    theta = math.radians(angle) # 弧度與角度轉換有專門的函數名稱，可以省略數學轉換。

    arc_distance = radius * theta
    chord_distance = 2 * radius * math.sin(theta / 2)

    print(f"{arc_distance:.6f} {chord_distance:.6f}")
```

注意:
1. 小數點位數寫法要會。
2. 弧度與角度轉換有專門的函數處理，無須手動數學轉換。
3. == 不可以寫成 is， == 是判斷物件內容值是否相同，而 is 是判斷兩個變數是否指向同一物件。
