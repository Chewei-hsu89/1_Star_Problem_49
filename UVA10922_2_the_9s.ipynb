# 2 the 9s

Time Limit: 1 sec

Description

A well-known trick to know if an integer N is a multiple of nine is to compute the sum S of its digits. If S is a multiple of nine, then so is N. This is a recursive test, and the depth of the recursion needed to obtain the answer on N is called the 9-degree of N.

Your job is, given a positive number N, determine if it is a multiple of nine and,if it is, its 9-degree.

Input

The input is a file such that each line contains a positive number. A line containing the number 0 is the end of the input. The given numbers can contain up to 1000 digits.

Output

The output of the program shall indicate, for each input number, if it is a multiple of nine, and in case it is, the value of its nine-degree. See the sample output for an example of the expected formatting of the output.


--------------------------------------------------------------------------------



```python
def degree_sum(x): # x is type int 
	r_seq = []
	while True:
		r = x % 10
		x = x // 10 # //是取商數，但若是除以負數，其結果為向下取整。
		r_seq.append(r)
		if x == 0:
			break
	k = sum(r_seq)
	return k
			 

while True:
	num = int(input())
	if num == 0:
		break
	
	if num % 9 != 0:
		print(num,"is not a multiple of 9.")
		
	else:
		degree_times = 0
		l = num
		while True:
			l = degree_sum(l)
			degree_times += 1
			if l == 9:
				break
		print(num,f"is a multiple of 9 and has 9-degree {degree_times}.")	
```

```python
# 精修版

# 直接採用累加，不需多弄list再一個個加入，更為精簡。
def digit_sum(x):
    total = 0

    while x > 0:
        total += x % 10
        x //= 10

    return total


while True:
    num = int(input())

    if num == 0:
        break

    if num % 9 != 0:
        print(f"{num} is not a multiple of 9.")
    else:
        degree_times = 0
        current = num # l 改為 current 會更具文字意義與辨別度

        while True:
            current = digit_sum(current)
            degree_times += 1

            if current == 9:
                break
        
        # print盡量寫在同一個f-string
        print(
            f"{num} is a multiple of 9 "
            f"and has 9-degree {degree_times}."
        )
```

```python
# GPT版本

while True:
    num = input().strip()

    if num == "0":
        break

    digit_sum = sum(map(int, num))

    if digit_sum % 9 != 0:
        print(f"{num} is not a multiple of 9.")
        continue #跳過當前的if，回到while

    degree = 1

    while digit_sum != 9:
        digit_sum = sum(map(int, str(digit_sum)))
        degree += 1

    print(f"{num} is a multiple of 9 and has 9-degree {degree}.")
```

小整理 & 補充:
1. break：直接結束整個迴圈；continue：只結束這一次，開始下一次迴圈；pass：什麼都不做，接著往下執行。
2. %: 取餘數； //: 對正整數取商數，對負整數向下取整； /: 一般除法，會有浮點數。
3. f-string寫法: print(f"jbaiufbwuebi{變數名稱}sifbuisbfse")。
4. map(int, num), num is string: 字串會一個一個字元做映射，比如map(int, "999")會取出"9","9","9"，若寫為map(int, ["999"])，則會取出"999"。
