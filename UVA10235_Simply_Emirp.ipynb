```python
### 自寫 ###
import math

## 判斷是否為質數，其中x僅處理大於2的部分。
def prime(x): #x is type int. x > 2.
	if x % 2 == 0:
		return 0 #0 represents not prime.
	else:
		for i in range(3, int(math.sqrt(x))+2, 2): #+2 is for justing the range
			if x % i == 0:
				return 0
		else:
			return 1

while True:
	try:
		num = int(input())
		
		if prime(num) == 0 and num > 2:
			print(f"{num} is not prime.")
		else:
			num_inversed = ""
			for j in reversed(str(num)): #利用str來來做反轉連接
				num_inversed += j
			if prime(int(num_inversed)) == 1 and num != int(num_inversed): #注意反轉的數字有可能與原本的數字相同
				print(f"{num} is emirp.")
			else:
				print(f"{num} is prime.")
				
	except EOFError:
		break
```

```python
### GPT版本 ###
import math
import sys


def is_prime(n):
    if n < 2:
        return False

    if n == 2:
        return True

    if n % 2 == 0:
        return False

    for divisor in range(3, math.isqrt(n) + 1, 2): #isqrt回傳整數
        if n % divisor == 0:
            return False

    return True


for line in sys.stdin:
    num = int(line)
    reversed_num = int(str(num)[::-1]) #用切片方式處理

    if not is_prime(num):
        print(f"{num} is not prime.")
    elif num != reversed_num and is_prime(reversed_num):
        print(f"{num} is emirp.")
    else:
        print(f"{num} is prime.")
```

Note:
1. 演算法的部分有多種:
    - 將num除以比它小的數字來判斷質數，注意只需除以到num的平方根即可。(基本)
    - 同前一個觀念，加入偶數部分剔除，僅處理奇數。(效率提升)
    - 所有大於3的質數可以表達成6k+1或6k-1。(高效)
2. math.sqrt() 與 math.isqrt() 前者為浮點數，後者為整數，無條件捨去小數位。
3. reverse相關用法:
    - .reverse()是list的用法，不能用在字串。
    - reversed(字串)是反向迭代器，
    - 與sort搭配，例如: list.sort(reverse=True)
    - 與sorted搭配，例如: new_numbers = sorted(numbers, reverse=True)
    - 字串版: reversed_text = "".join(reversed(text))
4. slice用法:
    - 型態: [start:stop:step]，例如[::-1]，start、stop沒寫代表適當的起點與終點，-1代表倒著走。
5. 布林值判斷說明:
    - if not is_prime(num):主要是看if所接收的，若接收為 if True，則判斷式執行；若接收為if False，則反之。
    - not僅是單純把布林值變號，而不是解讀為「是不是質數?」，而是單純看if接收的True or False。
