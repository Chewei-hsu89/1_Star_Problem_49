# D-1: 反轉字串

```python
import sys
from io import StringIO

raw_input = """You are handsome."""
sys.stdin = StringIO(raw_input)
s = sys.stdin.read().split()

print(s[::-1])
```

# D-2: ASCII

```python
import sys
from io import StringIO

raw_input = """Y"""
sys.stdin = StringIO(raw_input)
c = sys.stdin.read() #read完是str型態
print(ord(c))


n = 74
print(chr(n))
```
