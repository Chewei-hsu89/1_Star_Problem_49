# C-1: 統計頻率

輸入：

apple banana apple orange apple

輸出：

頻率

```python
import sys
from io import StringIO

raw_input = """apple banana apple orange apple"""
sys.stdin = StringIO(raw_input)

words = list(map(str, sys.stdin.read().split()))

freq = {} #建立dict

for w in words:
    freq[w] = freq.get(w, 0) + 1
    """
    freq.get(w, 0)
    去 dictionary freq 裡找 key=w
    找得到 → 回傳原本值
    找不到 → 回傳 0
    """

print(freq)
```

# C-2: Counter

```python
import sys
from io import StringIO
from collections import Counter

raw_input = """apple banana apple orange apple"""
sys.stdin = StringIO(raw_input)

words = list(map(str, sys.stdin.read().split())) #MAP多餘

c = Counter(words)

print(c)
print(c["apple"])
print(c.most_common(2)) #回傳前2名

for A, B in c.most_common(2):#輸出美觀
    print(A, B)
```

#　C-3: 找最多次

輸入：

1 1 2 2 2 3

輸出：

2

```python
import sys
from io import StringIO
from collections import Counter

raw_input = """1 1 2 2 2 3"""
sys.stdin = StringIO(raw_input)

n = list(map(int, sys.stdin.read().split()))

c = Counter(n)

print(c.most_common(2)[1][0]) #回傳前1名，[('apple', 3)]第一個[0]回傳整個list的，第二個[0]回傳tuple裡面的
print(c.most_common(1)[0][0])
```
