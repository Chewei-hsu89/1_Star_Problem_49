# B-1: 基本排序

輸入：

5 1 3 2 4

輸出排序後。

```python
import sys
from io import StringIO

raw_input = """5 1 3 2 4"""
sys.stdin = StringIO(raw_input)

arr = list(map(int, sys.stdin.read().split()))
arr.sort()
print(*arr)
```

# B-2: sorted vs sort

觀察 arr = [5,1,3] 在 b = sorted(arr) 與 arr.sort() 的差異。

【整理】
|           | `sort()`      | `sorted()`         |
| --------- | ------------- | ------------------ |
| 類型        | list method   | built-in function  |
| 是否修改原list | 會             | 不會                 |
| 回傳值       | `None`        | 新的list             |
| 使用方式      | `list.sort()` | `sorted(iterable)` |


```python
arr = [5,1,3]

b = sorted(arr)
print(*b)

arr.sort()
print(*arr)
```

# B-3: 奇偶排序（重要）

輸入：

5 2 9 1 8

輸出：

奇數在前，偶數在後。

```python
import sys
from io import StringIO

raw_input = """5 2 9 1 8"""
sys.stdin = StringIO(raw_input)

arr = list(map(int, sys.stdin.read().split()))

arr.sort(key=lambda x: x % 2) #lambda為匿名函式，架構: (key = lambda 參數: 回傳值)

print(*arr)
```

# B-4: 雙條件排序

續B-3，先 mod 3，再小到大

```python
import sys
from io import StringIO

raw_input = """5 2 9 1 8"""
sys.stdin = StringIO(raw_input)

arr = list(map(int, sys.stdin.read().split()))

arr.sort(key=lambda x: (x % 2, x)) #先比較x%2，再比較x

print(*arr)
```
