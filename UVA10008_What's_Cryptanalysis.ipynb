### What's Cryptanalysis?


```python
### 自寫 ###
import sys

data = sys.stdin.read() #type str

capital = []
count = 0
for i in data:
	if 64 < ord(i) < 91:
		capital.append(i)
	if  96 < ord(i) < 123:
		capital.append(chr(ord(i)-32))

tuples = []
for j in set(capital):
	num = capital.count(j)
	tuples.append((j, num))

tuples = sorted(tuples, key = lambda x : (-x[1], ord(x[0])))
for k in tuples:
	print(f"{k[0]} {k[1]}")
```

```python
### GPT版 ###
import sys
from collections import Counter

num_lines = int(sys.stdin.readline()) #讀第一行數字
counts = Counter()

for _ in range(num_lines):
    line = sys.stdin.readline().upper()
    counts.update(ch for ch in line if "A" <= ch <= "Z")

sorted_counts = sorted(counts.items(), key=lambda item: (-item[1], item[0]))

for letter, frequency in sorted_counts:
    print(f"{letter} {frequency}")
```

Note:
1. 原本寫法與Tell me the frequencies概念語法相似。
2. .upper()將所有文字大寫；.lower()將文字小寫。用在字串。
3. ch for ch in line if "A" <= ch <= "Z" 這邊特別注意字串也可以直接用A~Z，他們背後利用自己的Unicode編碼來比較。
4. counts 是 Counter 物件。它的 .update() 會把新資料的出現次數，累加到原本的統計中，而且第二次呼叫，先前的統計紀錄仍在，會持續累計。
